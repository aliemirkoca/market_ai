## Bu Projede Ne Yapmak İstiyorum?


Öncelikle bu projedeki niyetim kullanıcının verdiği yemek adının öncelikle tarifini bulup sonra yanına ne güzel gider onu bulup sonrasında bu yemeklerin tarifini ve malzemelerini bulan ondan sonra ise malzemeleri özel bi search modeli ile bulup kullanıcının önüne getiren bi yapay zeka modeli ve bunu göstermek için basit bi arayüz 

## Süreç Planım 

### Data Hazırlığı
Öncelikle Yemek adından tarif ve malzeme çıkarma kısmı için bana bolca yemek tarifi datası lazım ve o tarifleri ayrıştırıp kullanıcı önüne malzeme getirmek için güzel bi market database'ine ihtiyacım var .

### Reçete Çözümleme
İkinci adım olarak bu tarifleri işleyip ayrıştırmam lazım NLP ve Regex gibi yöntemlere başvuracağım

### Apriori Birliktelik
Burası biraz ekstra açıkcası güzel bi fikir katmak istedim bir yemeğin yanına gidebilecek meze yada karnıyarığın yanına pilav gider hadi bunu da yap gibisinden bir tavsiye motoru. Bu tavsiyeler verilirken satın alınacak malzemeler dikkate alınacak

### Anlamsal Eşleştirme Ve Optimizasyon
Malzemelerin Seçilerek Kullanıcıya sunulması ve sepete eklemesi için onaylanması kısmı burada tf-idf ve cosine similarity gibi yöntemlere başvurulacak

### Arayüz 
Açıkcası arayüz yazmak pek benim işim değil o yüzden basit bir şeyler yazıp bırakacağım amaç sistemi test etmek :)
